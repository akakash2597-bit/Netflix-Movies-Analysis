import matplotlib.pyplot as plt: Imports Python's core plotting library under the alias `plt` to enable custom chart generation.

plot(kind="bar") & plt.show(): Generates a bar graph with custom labels and titles, then displays the final visual layout on screen.


```python
import matplotlib.pyplot as plt
df["listed_in"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Genres on Netflix")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```

df["country"].value_counts().head(10).plot(kind="bar") : Counts occurrences per country, takes the top 10, and plots a bar chart.

```python
import matplotlib.pyplot as plt
df["country"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries with Most Content on Netflix")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```
